# Vision Data EDA

This notebook explores the **welding seam image dataset** used for YOLOv5 training.  
Goals:
- Inspect dataset structure
- Visualize sample images
- Explore label distribution
- Check class balance and bounding box stats


```python
# Imports
import os
import random
from pathlib import Path
import matplotlib.pyplot as plt
import cv2

# Dataset paths
images_dir = Path("../data/CV_Seam_Detection_Dataset/images")
labels_dir = Path("../data/CV_Seam_Detection_Dataset/labels")
```

## Dataset Overview


```python
image_files = list(images_dir.glob("*.jpg")) + list(images_dir.glob("*.png"))
label_files = list(labels_dir.glob("*.txt"))

print(f"Total images: {len(image_files)}")
print(f"Total labels: {len(label_files)}")
```

## Class Distribution


```python
from collections import Counter

class_counts = Counter()
for lf in label_files:
    with open(lf) as f:
        for line in f:
            class_id = line.strip().split()[0]
            class_counts[class_id] += 1

class_counts
```

```python
plt.bar(class_counts.keys(), class_counts.values())
plt.title("Class Distribution")
plt.xlabel("Class ID")
plt.ylabel("Count")
plt.show()
```

## Bounding Box Size Distribution


```python
widths, heights = [], []
for lf in label_files:
    with open(lf) as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) >= 5:
                _, _, _, w, h = parts[:5]
                widths.append(float(w))
                heights.append(float(h))

plt.hist(widths, bins=30, alpha=0.6, label="width")
plt.hist(heights, bins=30, alpha=0.6, label="height")
plt.legend()
plt.title("Bounding Box Size Distribution (normalized)")
plt.show()
```
