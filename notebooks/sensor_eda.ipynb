# Sensor Data EDA
This notebook explores the **sensor dataset** used for predictive maintenance.  
Goals:
- Inspect structure and quality of the CSV data
- Understand feature distributions
- Identify correlations and anomalies
- Prepare insights for model training

```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Settings
sns.set(style="whitegrid")
```

```python
# Load dataset
df = pd.read_csv("../data/welding_dataset.csv")
df.head()
```

## Basic Info

```python
df.info()
df.describe().T
```

## Missing Values


```python
df.isna().sum()
```

## Feature Distributions


```python
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols].hist(figsize=(15, 10), bins=30)
plt.suptitle("Sensor Feature Distributions")
plt.show()
```

## Correlation Heatmap


```python
plt.figure(figsize=(12,8))
sns.heatmap(df[numeric_cols].corr(), cmap="coolwarm", annot=False)
plt.title("Correlation Heatmap")
plt.show()
```

## Target Distribution (Defect column if present)


```python
if "Defect" in df.columns:
    sns.countplot(x="Defect", data=df)
    plt.title("Defect Distribution")
    plt.show()
```

## Outlier Detection (Boxplots)


```python
for col in numeric_cols[:6]:  # show first 6 for brevity
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

### Observations
- Note skewed features
- Identify highly correlated sensors
- Check class imbalance in Defect
